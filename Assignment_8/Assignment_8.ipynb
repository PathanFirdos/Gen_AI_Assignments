```python
import torch
import numpy as np
import soundfile as sf

from transformers import (
    SpeechT5Processor,
    SpeechT5ForTextToSpeech,
    SpeechT5HifiGan
)

# Load models
processor = SpeechT5Processor.from_pretrained(
    "microsoft/speecht5_tts"
)

model = SpeechT5ForTextToSpeech.from_pretrained(
    "microsoft/speecht5_tts"
)

vocoder = SpeechT5HifiGan.from_pretrained(
    "microsoft/speecht5_hifigan"
)

# Input text
text = "Hello this is text to speech using tacotron model"

inputs = processor(
    text=text,
    return_tensors="pt"
)

# Random speaker embedding (works for lab)
speaker_embeddings = torch.randn(1, 512)

# Generate speech (FINAL audio, not mel)
speech = model.generate_speech(
    inputs["input_ids"],
    speaker_embeddings,
    vocoder=vocoder
)

# Save file
sf.write(
    "speech.wav",
    speech.numpy(),
    samplerate=16000
)

print("Saved speech.wav")
```
