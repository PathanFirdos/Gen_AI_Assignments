Helper Function to Call the Model<br>This function sends a prompt to the GPT model and returns the generated response.

```python
def ask_gpt(prompt, model="gpt-4o-mini"):
    if DEMO_MODE:
        # -------- Simulated Responses (For Submission Without API) --------
        if "Classify the sentiment" in prompt:
            return "Positive"
        elif "best day ever" in prompt:
            return "Positive"
        elif "declining sales" in prompt:
            return "1. Sales decline due to reduced demand.\n2. Factors: competition, pricing, marketing.\n3. Collect sales and customer data.\n4. Recommend strategy revision."
        elif "Let's think step by step" in prompt:
            return "5 books x $20 = $100.\n10% of 100 = $10.\nFinal price = $90."
        elif "expand internationally" in prompt:
            return "Strategy A: Exporting.\nStrategy B: Franchising.\nStrategy C: Joint Venture.\nBest: Joint Venture due to shared risk."
        else:
            return "Demo response generated."

    # -------- Real API Call --------
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.7
        )
        return response.choices[0].message.content
    except Exception as e:
        print("Error occurred:", e)
        return "Request failed. Please check your API usage and billing."

```

### 3. ZERO-SHOT PROMPTING <br>                                                 Zero-shot prompting means asking the model to perform a task without giving any examples.


```python
zero_shot_prompt = """
Classify the sentiment of this sentence as Positive, Negative, or Neutral:
'I absolutely love this product. It works perfectly!'
"""

zero_shot_result = ask_gpt(zero_shot_prompt)
print("ZERO-SHOT RESULT:\n", zero_shot_result)
```

# 4. FEW-SHOT PROMPTING                                                        
Few-shot prompting provides examples before asking the model for a new prediction.

```python
few_shot_prompt = """
Classify sentiment as Positive, Negative, or Neutral.

Sentence: I hate waiting in long lines.
Sentiment: Negative

Sentence: The weather is okay today.
Sentiment: Neutral

Sentence: This is the best day ever!
Sentiment:
"""

few_shot_result = ask_gpt(few_shot_prompt)
print("\nFEW-SHOT RESULT:\n", few_shot_result)

```

# 5. INTERVIEW APPROACH                                                   
The Interview Approach involves breaking a task into interactive questions like an interviewer guiding reasoning step-by-step.

```python
interview_prompt = """
We are solving a business problem.

Question 1: What is the main issue in declining sales?
Question 2: What possible factors could contribute to it?
Question 3: What data would you collect?
Question 4: Provide a strategic recommendation.

Answer step by step.
"""

interview_result = ask_gpt(interview_prompt)
print("\nINTERVIEW APPROACH RESULT:\n", interview_result)
```

# 6. CHAIN-OF-THOUGHT (CoT)                                              
Chain-of-Thought prompting explicitly instructs the model to reason step by step.

```python
cot_prompt = """
If a store sells 5 books for $20 each and gives a 10% discount,
what is the total price after discount?

Let's think step by step.
"""

cot_result = ask_gpt(cot_prompt)
print("\nCHAIN-OF-THOUGHT RESULT:\n", cot_result)
```

# 7. TREE-OF-THOUGHT (ToT)                                  
Tree-of-Thought prompting explores multiple reasoning paths before selecting the best answer.

```python
tot_prompt = """
A company wants to expand internationally.

Step 1: Propose three different expansion strategies.
Step 2: Evaluate pros and cons of each.
Step 3: Choose the best strategy with justification.

Explore multiple possibilities before deciding.
"""

tot_result = ask_gpt(tot_prompt)
print("\nTREE-OF-THOUGHT RESULT:\n", tot_result)
```
