Step 2 Generate frames

```python
import cv2
import numpy as np
from PIL import Image
from IPython.display import HTML

# Load your image
person = Image.open("/content/stickman img.jpg").convert("RGBA")
person = person.resize((100, 100))  # scale image

width, height = 600, 400
frames = []

# Generate frames
for i in range(80):
    bg = Image.new("RGB", (width, height), "white")
    x = 10 + i * 5
    y = 250
    bg.paste(person, (x, y), person)  # paste with transparency
    frame = cv2.cvtColor(np.array(bg), cv2.COLOR_RGB2BGR)
    frames.append(frame)

# Save as MP4
video = cv2.VideoWriter("animation.mp4", cv2.VideoWriter_fourcc(*'mp4v'), 10, (width, height))

for f in frames:
    video.write(f)

video.release()
print("Saved animation.mp4")
```
